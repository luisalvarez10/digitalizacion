```python
def ph_agua():

  #Pedimos al usuario el valor del pH
  while True:
    try:
      ph = float(input("Introduce el pH del agua: "))
      break
    except ValueError:
      print("ERROR. INTRODUCE UN VALOR VÁLIDO\n")

  #Calculamos la acidez del agua
  if 0 <= ph <= 4:
    print("Es muy ácido.")

  elif 4 < ph < 7:
    print("Es poco ácido.")

  elif ph == 7:
    print("Es neutro.")

  elif 7 < ph <= 10:
    print("Es poco alcalino")

  elif 10 < ph <= 14:
    print("Es muy alcalino")


ph_agua()


```
